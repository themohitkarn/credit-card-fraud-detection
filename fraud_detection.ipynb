```python
import pandas as pd 
df = pd.read_csv('creditcard.csv')
df.head()
```

```python
df.info()
```

```python
df['Class'].value_counts()
```

```python
x=df.drop(columns=['Class'])
```

```python
y=df['Class']
```

```python
y
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.2,stratify=y,random_state=42)
```

```python
###Baseline model
from sklearn.linear_model import LogisticRegression  
from sklearn.metrics import classification_report
lr = LogisticRegression(max_iter=1000)
lr.fit(x_train,y_train)

y_pred_lr = lr.predict(x_test)
print(classification_report(y_test,y_pred_lr))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf=RandomForestClassifier(n_estimators=100,random_state=42,class_weight='balanced')
rf.fit(x_train, y_train)
```

```python
from sklearn.metrics import precision_score, recall_score ,f1_score
y_pred_rf=rf.predict(x_test)
```

```python
print("Precision", precision_score(y_test,y_pred_rf))
print("Recall", recall_score(y_test,y_pred_rf))
print("F1", f1_score(y_test,y_pred_rf))

print(classification_report(y_test,y_pred_rf))
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_res, y_res = smote.fit_resample(x_train, y_train)
```

```python
rf = RandomForestClassifier(
    n_estimators=500,
    max_depth=15,
    min_samples_split=5,
    min_samples_leaf=2,
    max_features="sqrt",
    class_weight="balanced",
    random_state=42
)
rf.fit(x_train, y_train)
```

```python
y_probs = rf.predict_proba(x_test)[:,1]

y_pred_custom = (y_probs > 0.3).astype(int)
```

```python
from sklearn.metrics import roc_auc_score

roc_auc_score(y_test, y_probs)
```

```python
from xgboost import XGBClassifier

xgb = XGBClassifier(
    scale_pos_weight=ratio,  # fraud / non-fraud
    n_estimators=300,
    max_depth=6,
    learning_rate=0.1
)
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv = StratifiedKFold(n_splits=5)
scores = cross_val_score(rf, x, y, cv=cv, scoring="f1")
```

```python
neg, pos = y_train.value_counts()
scale_pos_weight = neg / pos
scale_pos_weight

```

```python
from xgboost import XGBClassifier

xgb = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    random_state=42,
    eval_metric="logloss"
)

xgb.fit(x_train, y_train)
```

```python
from sklearn.metrics import classification_report, roc_auc_score

y_probs_xgb = xgb.predict_proba(x_test)[:,1]
y_pred_xgb = (y_probs_xgb > 0.3).astype(int)

print(classification_report(y_test, y_pred_xgb))
print("ROC-AUC:", roc_auc_score(y_test, y_probs_xgb))
```

```python
import joblib
joblib.dump(xgb, "xgboost_fraud_model.pkl")
```
